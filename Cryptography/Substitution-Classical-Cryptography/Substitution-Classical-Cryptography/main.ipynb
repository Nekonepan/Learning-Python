```python
def enkripsi():
  plaintext = input("MASUKKAN PESAN")
  k = int(input("Masukkan jumlah pergeseran (0-25): "))
  chipertext = ""
  for c in plaintext:
      if c.isalpha():
          c = c.upper()
          c = ord(c) - 65
          c = (c + k) % 26
          c = chr(c + 65)
      chipertext += c
  print("Chipertext : ", chipertext)

def dekripsi():
  chipertext = input("Masukkan chipertext: ")
  k = int(input("Masukkan jumlah pergeseran (0-25): "))
  plaintext = ""
  for c in chipertext:
    if c.isalpha():
      c = c.upper()
      c = ord(c) - 65
      if c- k < 0:
        c = 26 + (c - k)
      else:
        c = (c - k) % 26
      c = chr(c + 65)
      c = c.lower()
    plaintext += c
  print("Plaintext: ", plaintext)

def main():
  stop = False
  while not stop:
    print("MENU")
    print("1. ENKRIPSI")
    print("2. DEKRIPSI")
    print("3. EXIT")
    pil = int(input("Pilih menu : "))
    if pil == 1:
      enkripsi()
    elif pil == 2:
      dekripsi()
    elif pil == 3:
      stop = True

if __name__ == "__main__":
  main()
```
